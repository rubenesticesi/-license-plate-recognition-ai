# FASE 3: Exploración del dataset\nEn este notebook exploraremos el dataset ALPR descargado de Kaggle.\n

```
import os\nimport glob\nimport matplotlib.pyplot as plt\nimport cv2\n\n# Rutas base\nDATA_DIR = '../data/raw'\nTRAIN_DIR = os.path.join(DATA_DIR, 'train')\nVAL_DIR = os.path.join(DATA_DIR, 'valid')\nTEST_DIR = os.path.join(DATA_DIR, 'test')\n
```

```
def count_files(directory, ext='*.jpg'):\n    return len(glob.glob(os.path.join(directory, 'images', ext)))\n\nprint(f'Imágenes Train: {count_files(TRAIN_DIR)}')\nprint(f'Imágenes Valid: {count_files(VAL_DIR)}')\nprint(f'Imágenes Test: {count_files(TEST_DIR)}')\n
```

## Visualización de ejemplo\n

```
sample_image_path = glob.glob(os.path.join(TRAIN_DIR, 'images', '*.jpg'))[0]\nimg = cv2.imread(sample_image_path)\nimg = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)\n\nplt.imshow(img)\nplt.title('Ejemplo de Imagen ALPR')\nplt.axis('off')\nplt.show()\n
```
